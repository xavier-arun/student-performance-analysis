```python
import pandas as pd
df = pd.read_csv("dataset/student-mat.csv",sep=';')

# Dataset overview

print("Total number of students :",len(df))
print("\nTotal number of columns :",len(df.columns))
print("\nColumn names:", df.columns)
print("\nData types :",df.dtypes)
print("\nFirst 5 records :",df.head())
print("\nLast 5 records :",df.tail())
print("\nStatistical summary :",df.describe())
```

```python
#  Mathematics grade analysis
# Analyse the three grade columns :

def grade_analyse(grade) :
    print("\n   Average :",df[grade].mean())
    print("\n   Highest mark :",df[grade].max())
    print("\n   Lowest mark :",df[grade].min())
    print("\n   Median :",df[grade].median())
    print("\n   Standard deviation :",df[grade].std())

print("\nfor the first period-grade :")
grade_analyse("G1")
print("\nfor the second-period grade ")
grade_analyse("G2")
print("\nfor the final grade ")
grade_analyse("G3")
```

```python
#  Student performance analysis 

print("Top 10 students based on G3 :\n")
print(df.sort_values("G3", ascending=False).head(10))

print("\nStudents scoring above 15 :\n")
print(df[df["G3"] > 15])

print("\nStudents scoring below 10 :\n")
print(df[df["G3"] < 10])

print("\nStudents scoring exactly 10 :\n")
print(df[df["G3"] == 10])

print("\nHighest-performing student :\n")
print(df.iloc[df["G3"].idxmax()])

print("\nLowest-performing student :\n")
print(df.iloc[df["G3"].idxmin()])

print("\nOverall average final grade :\n")
print(df["G3"].mean())
```

```python
#  Study habits and performance
# this column showws how these factors relate to Mathematics grades:

print("the students who studied less than or equal to 2 hours got the average grade of :",df[df["studytime"] <= 2]["G3"].mean())
print("the students who studied more than 2 hours got the average grade of :",df[df["studytime"] > 2]["G3"].mean())

print("\nStudents with 0 to 2 past failures have the average G3:", df[df["failures"] <= 2]["G3"].mean())
print("Students with more than 2 past failures have the average G3:", df[df["failures"] > 2]["G3"].mean())

print("\nStudents with 0 to 25 absences have the average G3:", df[df["absences"] <= 25]["G3"].mean())
print("Students with 26 to 49 absences have the average G3:", df[(df["absences"] > 25) & (df["absences"] < 50)]["G3"].mean())
print("Students with 50 or more absences have the average G3:", df[df["absences"] >= 50]["G3"].mean())

print("\nthe students who has internet access has the average grade of :",df[df["internet"] == "yes"]["G3"].mean())
print("the students who has no internet access has the average grade of :",df[df["internet"] == "no"]["G3"].mean())

```

```python
# Data manipulation
print("Filter students based on marks :\n")
print("students who scored above 15 in G3 :\n",df[df["G3"] > 15])
print("\nstudents who scored below 10 in G3 :\n",df[df["G3"] < 10])
print("\nDisplay students who scored exactly 10 in G3 :\n",df[df["G3"] == 10])
print("\nstudents who scored between 10 and 15 in G3 :\n",df[(df["G3"] > 10) & (df["G3"] < 15)])

print("Sort students by final grade :")

print("\nstudents in G3 by ascending order :\n",df.sort_values("G3" , ascending= True ))
print("\nstudents in G3 by descending order :\n",df.sort_values("G3" , ascending= False ))
print("\nthe top 10 students based on G3 :\n",df.sort_values("G3", ascending = False).head(10))
print("\n the bottom 10 students based on G3 :\n",df.sort_values("G3", ascending = False).tail(10))

# to create new column :

df["performance"] = None
df.loc[df["G3"] >= 16 , "performance"] = "Excellent"
df.loc[(df["G3"] == 14) | (df["G3"] == 15 ), "performance"] = "Good"
df.loc[(df["G3"] > 9) & (df["G3"] <= 13 ), "performance"] = "Average"
df.loc[df["G3"] <= 9, "performance"] = "Needs improvement"

df





```
